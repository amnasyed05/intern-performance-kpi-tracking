```python
import pandas as pd

data = pd.read_csv('intern_kpi_data.csv')
data.columns = data.columns.str.strip()
print(data.columns.tolist())
data.head()
```

```python
data['CompletionScore'] = 100 - (data['AVERAGE COMPLETION TIME'] * 8)
data['CompletionScore'] = data['CompletionScore'].clip(lower=0)

data['OverallScore'] = (
    data['CompletionScore'] * 0.3 +
    data['QUALITY SCORE'] * 10 * 0.4 +
    data['MENTOR FEEDBACK'] * 20 * 0.3
)

print(data[['NAME','MONTH','OverallScore']])
```

```python
monthly_report = data.groupby('MONTH').agg({
    'AVERAGE COMPLETION TIME': 'mean',
    'QUALITY SCORE': 'mean',
    'MENTOR FEEDBACK': 'mean',
    'OverallScore': 'mean'
}).reset_index()

print(monthly_report)
```

```python
# Top 3 performers each month
top_performers = data.sort_values(['MONTH','OverallScore'], ascending=[True, False]).groupby('MONTH').head(3)
print("TOP PERFORMERS:")
print(top_performers[['NAME','MONTH','OverallScore']])

# Bottom 3 performers each month (may need support)
bottom_performers = data.sort_values(['MONTH','OverallScore'], ascending=[True, True]).groupby('MONTH').head(3)
print("\nNEEDS SUPPORT:")
print(bottom_performers[['NAME','MONTH','OverallScore']])
```

```python
monthly_report.to_csv('monthly_kpi_report.csv', index=False)
data.to_csv('intern_kpi_scored_data.csv', index=False)
```
